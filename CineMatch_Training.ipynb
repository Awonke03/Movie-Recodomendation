# CineMatch v2: chronological recommendation evaluation
This notebook runs the same training pipeline used by the application. Dataset download requires network access. No test-set tuning.

```python
import sys, subprocess, json
from pathlib import Path
subprocess.run([sys.executable,'train_recommender.py'],check=True)
report=json.loads(Path('model/report.json').read_text())
```

```python
import pandas as pd
display(pd.DataFrame(report['validation']).T.sort_values('ndcg_at_10',ascending=False))
display(pd.DataFrame(report['test']).T.sort_values('ndcg_at_10',ascending=False))
print(report['champion'],report['ndcg_gain_ci_vs_popularity'],report['test_population'])
```

```python
from recommender import load_artifacts,rank
model,report=load_artifacts('model')
seeds=[i for i,m in enumerate(model['movies']) if m['title'] in ['Toy Story','Toy Story 2','Shrek']]
display(pd.DataFrame([model['movies'][i] for i in rank(model,seeds,limit=12)]))
```
